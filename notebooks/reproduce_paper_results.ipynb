# Reproduce the empirical paper outputs

This notebook regenerates current tables and figures from the compact frozen paper metrics. It does not train models, solve OT, download private artifacts, or load participant-level records. Output names are semantic and independent of manuscript numbering.

```python
from pathlib import Path
import subprocess
import sys
import pandas as pd
from IPython.display import Image, Markdown, display

def find_project_root(start=Path.cwd()):
    for candidate in (start, *start.parents):
        if (candidate / 'src/otrank').is_dir() and (candidate / 'paper_data/manifest.json').is_file():
            return candidate
        if (candidate / 'code/src/otrank').is_dir() and (candidate / 'code/paper_data/manifest.json').is_file():
            return candidate
    raise FileNotFoundError('Could not locate the project root')

PROJECT = find_project_root()
CODE = PROJECT if (PROJECT / 'src/otrank').is_dir() else PROJECT / 'code'
subprocess.run([sys.executable, str(CODE / 'reproduce_paper_outputs.py'), '--project-root', str(PROJECT)], check=True)
print(PROJECT)
```

## Main-text empirical tables

```python
linear = PROJECT / 'results/simulation/linear_n1000_p20_d10_final'
relu = PROJECT / 'results/simulation/relu_n1000_p8_d4_final'
nhanes = PROJECT / 'results/real_data/nhanes_p43_final'

display(Markdown('### Linear estimation and equivariance'))
display(pd.read_csv(linear / 'summaries/point_estimation_summary.csv'))
display(Markdown('### Linear confidence intervals'))
display(pd.read_csv(linear / 'summaries/inference_summary.csv'))
display(Markdown('### NHANES results'))
display(pd.read_csv(nhanes / 'tables/nhanes_table.csv'))
```

## Simulation figures

```python
display(Markdown('### Coefficient error and equivariance'))
display(Image(filename=str(linear / 'figures/linear_accuracy_equivariance_boxplots.png')))
display(Markdown('### Confidence-interval length and coverage'))
display(Image(filename=str(linear / 'figures/linear_inference_boxplots.png')))
```

## Supplementary tables

```python
display(Markdown('### ReLU prediction'))
display(pd.read_csv(relu / 'summaries/prediction_summary.csv'))
display(Markdown('### Retained NHANES predictors'))
display(pd.read_csv(nhanes / 'tables/nhanes_predictors.csv'))
display(Markdown('### Pairwise representation stability'))
display(pd.read_csv(nhanes / 'summaries/pairwise_stability.csv'))
display(Markdown('### Semi-synthetic inference stability'))
display(pd.read_csv(nhanes / 'supplement/semisynthetic_invariance/comparison/summary.csv'))
```
