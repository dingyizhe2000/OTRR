# Reproduce the empirical paper outputs

This notebook rebuilds and displays outputs from the three retained paper domains: the linear simulation, the ReLU simulation, and the NHANES application with its supplementary experiments. In the current manuscript numbering, the empirical main tables are Tables 3--5; Table 2 is theoretical and is not generated here.

```python
from pathlib import Path
import subprocess
import sys
import pandas as pd
from IPython.display import Image, Markdown, display

def find_project_root(start=Path.cwd()):
    for candidate in (start, *start.parents):
        if (candidate / 'src/otrank').is_dir() and (candidate / 'results').is_dir():
            return candidate
        if (candidate / 'code/src/otrank').is_dir() and (candidate / 'results').is_dir():
            return candidate
    raise FileNotFoundError('Could not locate the project root')

PROJECT = find_project_root()
CODE = PROJECT if (PROJECT / 'src/otrank').is_dir() else PROJECT / 'code'
subprocess.run([sys.executable, str(CODE / 'reproduce_paper_outputs.py'), '--project-root', str(PROJECT)], check=True)
print(PROJECT)
```

## Main-text empirical tables

```python
linear = PROJECT / 'results/simulation/linear_n1000_p20_d10_final'
relu = PROJECT / 'results/simulation/relu_n1000_p8_d4_final'
nhanes = PROJECT / 'results/real_data/nhanes_p43_final'

display(Markdown('### Table 3: linear estimation and invariance'))
display(pd.read_csv(linear / 'summaries/point_estimation_summary.csv'))
display(Markdown('### Table 4: linear inference'))
display(pd.read_csv(linear / 'summaries/inference_summary.csv'))
display(Markdown('### Table 5: NHANES'))
display(pd.read_csv(nhanes / 'tables/table5.csv'))
```

## Supplementary Figures S2.1 and S2.2

```python
display(Markdown('### Figure S2.1'))
display(Image(filename=str(linear / 'figures/linear_accuracy_invariance_boxplots.png')))
display(Markdown('### Figure S2.2'))
display(Image(filename=str(linear / 'figures/linear_inference_boxplots.png')))
```

## Supplementary tables

```python
display(Markdown('### Table S2.1: ReLU prediction'))
display(pd.read_csv(relu / 'summaries/prediction_summary.csv'))
display(Markdown('### Table S2.2: retained NHANES predictors'))
display(pd.read_csv(nhanes / 'tables/table_s2_2_predictors.csv'))
display(Markdown((nhanes / 'tables/table_s2_2_STATUS.md').read_text()))
display(Markdown('### Table S2.3: pairwise representation stability'))
display(pd.read_csv(nhanes / 'summaries/pairwise_stability.csv'))
display(Markdown('### Table S2.4: semi-synthetic inference stability'))
display(pd.read_csv(nhanes / 'supplement/semisynthetic_invariance/comparison/summary.csv'))
```
